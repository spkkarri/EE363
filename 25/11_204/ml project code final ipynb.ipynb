```python
import scipy.io
import numpy as np
import torch
import torch.nn as nn
from torch.utils.data import DataLoader, TensorDataset
import matplotlib.pyplot as plt
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score, confusion_matrix
import seaborn as sns

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

# === Load data from .mat files ===
def load_soh_per_battery(filenames):
    soh_dict = {}
    for file in filenames:
        data = scipy.io.loadmat(file)
        battery_name = file.split(".")[0]
        battery = data[list(data.keys())[-1]][0][0]
        cycles = battery['cycle'][0]
        battery_soh = []
        for c in cycles:
            if c['type'][0] == 'discharge':
                cap = c['data'][0, 0]['Capacity'][0][0]
                battery_soh.append(cap)
        soh_dict[battery_name] = np.array(battery_soh, dtype=np.float32)
    return soh_dict

mat_files = ["B0005.mat", "B0006.mat", "B0007.mat", "B0018.mat"]
soh_data = load_soh_per_battery(mat_files)
all_soh = np.concatenate([soh_data[k] for k in soh_data])

# Normalize
soh_min, soh_max = all_soh.min(), all_soh.max()
soh_norm = (all_soh - soh_min) / (soh_max - soh_min)

# === Sequence creation ===
def create_sequences(data, seq_len):
    X, y = [], []
    for i in range(len(data) - seq_len):
        X.append(data[i:i + seq_len])
        y.append(data[i + seq_len])
    return np.array(X), np.array(y)

seq_len = 20
X, y = create_sequences(soh_norm, seq_len)

X_tensor = torch.tensor(X, dtype=torch.float32).to(device)
y_tensor = torch.tensor(y, dtype=torch.float32).to(device)
dataset = TensorDataset(X_tensor, y_tensor)
dataloader = DataLoader(dataset, batch_size=32, shuffle=True)

# === Models ===
class TransformerSOHPredictor(nn.Module):
    def __init__(self, seq_len, model_dim=64, num_layers=2, nhead=4, pred_len=1):
        super().__init__()
        self.input_projection = nn.Linear(seq_len, model_dim)
        self.positional_encoding = nn.Parameter(torch.zeros(1, seq_len, model_dim))
        encoder_layer = nn.TransformerEncoderLayer(d_model=model_dim, nhead=nhead, dim_feedforward=128)
        self.transformer_encoder = nn.TransformerEncoder(encoder_layer, num_layers=num_layers)
        self.decoder = nn.Linear(model_dim, pred_len)

    def forward(self, x):
        x = self.input_projection(x)
        x = x.unsqueeze(1) + self.positional_encoding[:, :1]
        x = x.permute(1, 0, 2)
        x = self.transformer_encoder(x)
        out = self.decoder(x[-1])
        return out.squeeze()

class LSTM_CNN_SOHPredictor(nn.Module):
    def __init__(self, seq_len, hidden_dim=64, cnn_out=32):
        super().__init__()
        self.lstm = nn.LSTM(input_size=1, hidden_size=hidden_dim, batch_first=True)
        self.conv = nn.Sequential(
            nn.Conv1d(1, 16, kernel_size=3, padding=1),
            nn.ReLU(),
            nn.Conv1d(16, cnn_out, kernel_size=3, padding=1),
            nn.ReLU(),
            nn.AdaptiveAvgPool1d(1)
        )
        self.fc = nn.Linear(hidden_dim + cnn_out, 1)

    def forward(self, x):
        x = x.unsqueeze(-1)
        lstm_out, _ = self.lstm(x)
        lstm_feat = lstm_out[:, -1, :]
        cnn_feat = self.conv(x.permute(0, 2, 1)).squeeze(-1)
        combined = torch.cat((lstm_feat, cnn_feat), dim=1)
        out = self.fc(combined)
        return out.squeeze()

# === Training & Evaluation ===
def train_model(model, dataloader, epochs=100, lr=0.001):
    model.to(device)
    optimizer = torch.optim.Adam(model.parameters(), lr=lr)
    criterion = nn.MSELoss()
    best_loss = float('inf')
    for epoch in range(epochs):
        model.train()
        total_loss = 0
        for xb, yb in dataloader:
            optimizer.zero_grad()
            pred = model(xb)
            loss = criterion(pred, yb)
            loss.backward()
            optimizer.step()
            total_loss += loss.item()
        avg_loss = total_loss / len(dataloader)
        print(f"Epoch {epoch+1}, Loss: {avg_loss:.4f}")
    return model

def evaluate(model, X_tensor, y_tensor, label):
    model.eval()
    with torch.no_grad():
        preds = model(X_tensor).cpu().numpy()
    true = y_tensor.cpu().numpy()
    pred_soh = preds * (soh_max - soh_min) + soh_min
    true_soh = true * (soh_max - soh_min) + soh_min

    mae = mean_absolute_error(true_soh, pred_soh)
    rmse = mean_squared_error(true_soh, pred_soh, squared=False)
    r2 = r2_score(true_soh, pred_soh)
    mape = np.mean(np.abs((true_soh - pred_soh) / true_soh)) * 100

    print(f"\n📈 {label} Evaluation:")
    print(f"MAE: {mae:.4f}, RMSE: {rmse:.4f}, R²: {r2:.4f}, MAPE: {mape:.2f}%")
    return true_soh, pred_soh

# === Train and Evaluate ===
transformer = train_model(TransformerSOHPredictor(seq_len), dataloader)
true_t, pred_t = evaluate(transformer, X_tensor, y_tensor, "Transformer")

lstmcnn = train_model(LSTM_CNN_SOHPredictor(seq_len), dataloader)
true_l, pred_l = evaluate(lstmcnn, X_tensor, y_tensor, "LSTM+CNN")

# === Plot comparison ===
plt.figure(figsize=(10, 5))
plt.plot(true_t, label='Actual SOH', color='black')
plt.plot(pred_t, label='Transformer', alpha=0.7)
plt.plot(pred_l, label='LSTM+CNN', alpha=0.7)
plt.title("📊 Model Comparison: SOH Prediction")
plt.xlabel("Cycle Index")
plt.ylabel("SOH")
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()

# === Residuals ===
residuals = true_t - pred_t
plt.figure(figsize=(10, 4))
plt.plot(residuals, label="Residuals", color="red")
plt.axhline(0, color='black', linestyle='--')
plt.title("🧮 Transformer Residual Plot")
plt.xlabel("Cycle Index")
plt.ylabel("Error")
plt.grid(True)
plt.tight_layout()
plt.show()

# === Scatter Plot ===
plt.figure(figsize=(6, 6))
plt.scatter(true_t, pred_t, alpha=0.6, color='green')
plt.plot([min(true_t), max(true_t)], [min(true_t), max(true_t)], 'k--')
plt.xlabel("Actual SOH")
plt.ylabel("Predicted SOH")
plt.title("📌 Actual vs Predicted SOH (Transformer)")
plt.grid(True)
plt.tight_layout()
plt.show()

# === Error Histogram ===
plt.figure(figsize=(6, 4))
plt.hist(residuals, bins=30, color="orange", edgecolor="black")
plt.title("📊 Error Distribution")
plt.xlabel("Residual (Actual - Predicted)")
plt.tight_layout()
plt.show()

# === Confusion Matrix (binning SOH) ===
bins = np.linspace(min(true_t), max(true_t), 10)
true_bin = np.digitize(true_t, bins)
pred_bin = np.digitize(pred_t, bins)
conf_matrix = confusion_matrix(true_bin, pred_bin)

plt.figure(figsize=(6, 5))
sns.heatmap(conf_matrix, annot=True, fmt='d', cmap='Blues')
plt.title("🧮 Confusion Matrix (Binned SOH)")
plt.xlabel("Predicted Bin")
plt.ylabel("Actual Bin")
plt.tight_layout()
plt.show()

# === Capacity Graphs Per Battery ===
plt.figure(figsize=(12, 8))
for battery, soh in soh_data.items():
    plt.plot(soh, label=battery)
plt.title("🔋 Capacity Curves for Each Battery")
plt.xlabel("Cycle Index")
plt.ylabel("Capacity (SOH)")
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
# === Visualization Function ===
def visualize_predictions(true_soh, pred_soh, model_name="Model"):
    residuals = true_soh - pred_soh

    # 1. Prediction vs Actual
    plt.figure(figsize=(10, 5))
    plt.plot(true_soh, label='Actual SOH', color='black')
    plt.plot(pred_soh, label=f'{model_name} Prediction', alpha=0.7)
    plt.title(f"📊 {model_name}: Prediction vs Actual")
    plt.xlabel("Cycle Index")
    plt.ylabel("SOH")
    plt.legend()
    plt.grid(True)
    plt.tight_layout()
    plt.show()

    # 2. Residual Plot
    plt.figure(figsize=(10, 4))
    plt.plot(residuals, label="Residuals", color="red")
    plt.axhline(0, color='black', linestyle='--')
    plt.title(f"🧮 {model_name}: Residual Plot")
    plt.xlabel("Cycle Index")
    plt.ylabel("Error")
    plt.grid(True)
    plt.tight_layout()
    plt.show()

    # 3. Scatter Plot
    plt.figure(figsize=(6, 6))
    plt.scatter(true_soh, pred_soh, alpha=0.6, color='green')
    plt.plot([min(true_soh), max(true_soh)], [min(true_soh), max(true_soh)], 'k--')
    plt.xlabel("Actual SOH")
    plt.ylabel("Predicted SOH")
    plt.title(f"📌 {model_name}: Actual vs Predicted")
    plt.grid(True)
    plt.tight_layout()
    plt.show()

  

    # 5. Confusion Matrix (Binned)
    bins = np.linspace(min(true_soh), max(true_soh), 10)
    true_bin = np.digitize(true_soh, bins)
    pred_bin = np.digitize(pred_soh, bins)
    conf_matrix = confusion_matrix(true_bin, pred_bin)

    plt.figure(figsize=(6, 5))
    sns.heatmap(conf_matrix, annot=True, fmt='d', cmap='Blues')
    plt.title(f"🧩 {model_name}: Confusion Matrix (Binned)")
    plt.xlabel("Predicted Bin")
    plt.ylabel("Actual Bin")
    plt.tight_layout()
    plt.show()

# === Plot for Transformer ===
visualize_predictions(true_t, pred_t, model_name="Transformer")

# === Plot for LSTM+CNN ===
visualize_predictions(true_l, pred_l, model_name="LSTM + CNN")
```
